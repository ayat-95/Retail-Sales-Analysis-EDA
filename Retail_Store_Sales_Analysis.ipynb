#  5. Data Cleaning 

### 5.1 convart the data column to appropriate data type

```python
# convert the transaction_date column to datetime form

df['Transaction Date']= pd.to_datetime(df['Transaction Date'])
df['Transaction Date']
```

### 5.2    fill missing values in the dataset 

```python
1 # Price (Total_spent/ Quantity )

df['Price Per Unit'] =df['Price Per Unit'].fillna(df['Total Spent']/df['Quantity'])
df['Price Per Unit'] .isnull().sum()
```

```python
2 # Quantity (Total_spent / price) wrong

# df['Quantity'] =df['Quantity'].fillna(df['Total Spent']/df['Price Per Unit'])
# df['Quantity'].isnull().sum()
```

```python
2 # Quantity 
median_Quantity = df['Quantity'].median().round(2)
df['Quantity']=df['Quantity'].fillna(median_Quantity)
```

```python
3 # Total Spent 
df['Total Spent'] =df['Total Spent'].fillna(df['Price Per Unit'] * df['Quantity'])
```

```python
4 # Discount Applied fill by Unknown(no diff between them)
df['Discount Applied']= df['Discount Applied'].fillna('UnKnown')
```

```python
5 # Item 
item_mapping = (
    df.dropna(subset=["Item", "Category", "Price Per Unit"])
    .groupby(["Category", "Price Per Unit"])["Item"]
    .first()
)


df["mapped_item"] = df.set_index(["Category", "Price Per Unit"]).index.map(
    item_mapping
)
df["Item"] = df["Item"].fillna(df["mapped_item"])


df.drop(columns=["mapped_item"], inplace=True)  


print("Done")
```

# 8.Feature Engineering

### - Extract year, month, day from transaction date

```python
# Create year column
df['Year']=df['Transaction Date'].dt.year
df['Year']
```

```python
# Create month_name column
df['Month']=df['Transaction Date'].dt.month_name()
df['Month']
```

```python
# Create day_name column

df['Day']=df['Transaction Date'].dt.day_name()
df['Day']
```

```python
# Extract weekday and weekend

def day_category(x):
    if x in ('Saturday','Sunday'):
        return 'weekend'
    else:
        return 'weekday'

df['Day_type'] = df['Day'].apply(day_category)
df['Day_type'] 
```

# Recommendations

# 9. Business Recommendations

Based on the analysis and the identified sales and customer patterns, the following recommendations can be considered:

### 1. Focus on High-Performing Product Categories

The **Butchers** category generated the highest revenue at **$218,153**. The business should closely monitor inventory and demand for high-performing categories and consider prioritizing them in marketing and sales strategies.

### 2. Strengthen the Online Sales Channel

**Online sales generated $831,922**, making Online the strongest sales channel in terms of revenue. The business could explore opportunities to improve the online shopping experience and encourage further online purchases.

### 3. Retain High-Value Customers

**CUST_24** generated **$71,890**, indicating the importance of high-value customers. The business should identify and retain similar high-value customers through personalized offers, loyalty strategies, and targeted communication.

### 4. Monitor Cash Payment Usage

**Cash was the most frequently used payment method, with 4,310 transactions.** The business should continue supporting cash payments while monitoring whether digital payment methods are growing or generating higher-value transactions.

### 5. Investigate Weekend Purchasing Behavior

Weekend transactions had the highest **average transaction value**, at approximately **$131.29**. The business could investigate the reasons behind higher weekend spending and consider targeted weekend promotions or product bundles.

### 6. Prioritize High-Revenue Products

**Item_25_FUR** generated the highest product revenue at **$27,101**. The business should monitor its availability and demand and consider using similar high-performing products in promotional and merchandising strategies.

### 7. Use Data-Driven Decision Making

Future business decisions should consider multiple indicators—including revenue, transaction frequency, average transaction value, customer contribution, and sales channel performance—rather than relying on a single metric.
